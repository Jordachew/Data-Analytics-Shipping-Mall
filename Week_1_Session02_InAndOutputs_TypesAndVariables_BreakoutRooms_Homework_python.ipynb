###Exercise: BMI Calculator
Write a program that calculates the BMI (Body Mass Index) for the user. Ask the user to provide his/her weight (kg) and height (cm).

How to calculate BMI:

BMI = weight_kg / (height_cm/100)²

Print the result as follows: print(f"Your BMI is {BMI}")

```python
weight_kg=input("What is your weight in kg? ")
height_cm=input("What is your height in cm? ")
BMI=float(weight_kg)/((float(height_cm)/100)**2)
print("Thank you for your input. Your BMI is",round(BMI,1))
```

###Exercise: Boolean Values and if-conditions

If the user is younger than 18 years, print the following:
