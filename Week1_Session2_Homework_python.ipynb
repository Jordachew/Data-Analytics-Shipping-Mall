If you run a 10 kilometer race in 42 minutes 42 seconds, what is your average pace (time per mile in minutes and seconds)?

```python
min_2=42
sec_2=42
km=10
k_mconvert=1.61
mile_conv=round(km/k_mconvert,2)
answer_min=int((min_2*60+int(sec_2)/60/60)/mile_conv/60)
answer_sec=round((((min_2*60+int(sec_2)/60/60)/mile_conv/60)-answer_min)*60,2)
print("Average pace=",answer_min,"min and",answer_sec)
```
